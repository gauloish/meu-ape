## Tratando e Limpando as Features

Agora, vamos analisar feature por feature, tratar e limpar seus valores.

```python
def get_info(column, values=True):
    num_na = column.isna().sum()
    num_total = len(column)

    print(f"dtype: {column.dtype}")
    print(f"NA: {num_na}/{num_total} ({100*num_na/num_total:.2f}%)")

    counts = column.value_counts(dropna=False)

    if values:
        df_values = pd.DataFrame({
            "contagem": counts,
            "porcentagem": 100*counts/num_total
        })

        df_values["porcentagem"] = (df_values["porcentagem"]
            .map(lambda x: f"{x:.2f} %")
        )
        
        print(f"valores:\n{df_values}")
```

### Feature `id`

```python
get_info(df["id"])
```

Como podemos ver, a feature apenas representa os id's dos imóveis. Como essa informação não estará presente durante a inferência, então podemos remover ela.

```python
df = df.drop("id", axis="columns")
```

### Feature `preco` (target)

```python
get_info(df["preco"])
```

Pelas informações acima podemos notar que o preço está codificado como texto com um padrão simples. Vamos checar se toda a coluna segue esse padrão.

```python
df[~df["preco"].str.match(r"^R[$] (\d+\.)*(\d+)+(,\d+)?")]["preco"]
```

Como podemos ver alguns valores não seguem o padrão e também não correspondem a um preço válido. Como não há valores faltantes então vamos eliminar esses registros que não seguem o padrão já que não podemos usâ-los para o treinamento.

```python
df = df[df["preco"].str.match(r"^R[$] (\d+\.)*(\d+)+(,\d+)?")]
```

Agora vamos extrair os valores numéricos e converter a coluna para um tipo numérico.

```python
df["preco"] = (df["preco"]
    .str.extract(r"^R[$] (.*)")[0]
    .str.replace(".", "")
)
```

```python
df["preco"].str.match(",").sum()
```

```python
df["preco"] = df["preco"].astype("float")
```

### Feature `condominio`

### Feature `tamanho`

```python
get_info(df["tamanho"])
```

Vamos checar se todos os registros seguem esse padrão de apresentação do tamanho do imóvel.

```python
df[~df["tamanho"].str.match(r"\d+ m²")]["tamanho"]
```

Como podemos ver, alguns registros têm uma faixa de valores do imóvel ao invés de um valor em si. Como são poucos registros, iremos manter apenas o maior valor dessas faixar de valores para obtermos um valor só.

```python
df["tamanho"] = (df["tamanho"]
    .str.replace(r"^\d+ - ", "", regex=True)
    .str.extract(r"^(.*) m²")[0]
    .str.replace(r"\.", "")
    .astype("float")
)
```
