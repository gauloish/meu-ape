## Tratando e Limpando as Features

Agora, vamos analisar feature por feature, tratar e limpar seus valores.

```python
def get_info(column, values=True):
    num_na = column.isna().sum()
    num_total = len(column)

    print(f"dtype: {column.dtype}")
    print(f"NA: {num_na}/{num_total} ({100*num_na/num_total:.2f}%)")

    counts = column.value_counts(dropna=False)

    if values:
        df_values = pd.DataFrame({
            "contagem": counts,
            "porcentagem": 100*counts/num_total
        })

        df_values["porcentagem"] = (df_values["porcentagem"]
            .map(lambda x: f"{x:.2f} %")
        )
        
        print(f"valores:\n{df_values}")
```

## Validando o Dataset

Nessa etapa iremos validar algumas informações do dataset com a feature `titulo`.

```python
quartos = df["titulo"].str.extract(r"(?P<quarto>\d+) quarto[s]")["quarto"].astype("float64")
banheiros = df["titulo"].str.extract(r"(?P<banheiro>\d+) banheiro[s]")["banheiro"].astype("float64")
vagas = df["titulo"].str.extract(r"(?P<vaga>\d+) vaga[s]")["vaga"].astype("float64")
area_m2 = df["titulo"].str.extract(r"(?P<area>\d+) m²")["area"].astype("float64")

df_validate = pd.DataFrame({
    "quartos": quartos,
    "banheiros": banheiros,
    "vagas": vagas,
    "area_m2": area_m2,
})

df_validate
```

```python
df_values = df[["quartos", "banheiros", "vagas", "area_m2"]].values.reshape(-1)
df_validate_values = df_validate.values.reshape(-1)
mask = pd.Series(df_validate_values).notna()

correct = (df_values[mask] == df_validate_values[mask]).sum()
total = mask.sum()

print(f"Porcentagem de dados coerentes: {100*correct / total:.2f}%")
```

Como podemos ver, das informações presentes no título do anúncio cerca de $97.98\%$ estão coerentes com os dados extraídos, o que representa uma alta taxa de integridade.
